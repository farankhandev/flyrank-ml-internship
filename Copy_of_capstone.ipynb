```python
import pandas as pd
import numpy as np


apr_gsc = apr.loc[apr["gsc_data_available"]].copy()
may_gsc = may.loc[may["gsc_data_available"]].copy()
jun_gsc = df.loc[df["gsc_data_available"]].copy()


def aggregate_month(data, prefix):
    result = (
        data.groupby(["client_hash_id", "content_hash_id"], as_index=False)
        .agg(
            impressions=("gsc_impressions", "sum"),
            clicks=("gsc_clicks", "sum"),
            days=("report_date", "nunique"),
            position_sum=("gsc_sum_position", "sum")
        )
    )
    return result.rename(columns={
        "impressions": f"{prefix}_impressions",
        "clicks": f"{prefix}_clicks",
        "days": f"{prefix}_days",
        "position_sum": f"{prefix}_position_sum"
    })

apr_agg = aggregate_month(apr_gsc, "apr")
may_agg = aggregate_month(may_gsc, "may")
jun_agg = aggregate_month(jun_gsc, "jun")


model_df = apr_agg.merge(
    may_agg,
    on=["client_hash_id", "content_hash_id"],
    how="outer"
).merge(
    jun_agg,
    on=["client_hash_id", "content_hash_id"],
    how="outer"
)


numeric_cols = model_df.select_dtypes(include="number").columns
model_df[numeric_cols] = model_df[numeric_cols].fillna(0)


model_df["hist_impressions"] = (
    model_df["apr_impressions"] + model_df["may_impressions"]
)
model_df["hist_clicks"] = model_df["apr_clicks"] + model_df["may_clicks"]
model_df["hist_days"] = model_df["apr_days"] + model_df["may_days"]


model_df["historical_30d_impressions"] = (
    model_df["hist_impressions"] / model_df["hist_days"].clip(lower=1) * 30
)


model_df["june_30d_impressions"] = (
    model_df["jun_impressions"] / model_df["jun_days"].clip(lower=1) * 30
)

# Eligibility rules
model_df = model_df[
    (model_df["hist_impressions"] >= 100)
    & (model_df["hist_days"] >= 40)
    & (model_df["jun_days"] >= 20)
].copy()


model_df["is_declining"] = (
    model_df["june_30d_impressions"]
    < 0.80 * model_df["historical_30d_impressions"]
).astype(int)

print("Modeling dataset shape:", model_df.shape)
print("\nTarget distribution:")
print(model_df["is_declining"].value_counts())
print("\nTarget percentages:")
print((model_df["is_declining"].value_counts(normalize=True) * 100).round(2))
```

```python
import duckdb
from huggingface_hub import hf_hub_download

repo = "FlyRank/internship-warehouse"

apr_path = hf_hub_download(
    repo_id=repo,
    filename="fact_content_daily_performance/month=2026-04/data_0.parquet",
    repo_type="dataset"
)
may_path = hf_hub_download(
    repo_id=repo,
    filename="fact_content_daily_performance/month=2026-05/data_0.parquet",
    repo_type="dataset"
)
jun_path = hf_hub_download(
    repo_id=repo,
    filename="fact_content_daily_performance_sample.parquet",
    repo_type="dataset"
)

con = duckdb.connect("/content/flyrank.duckdb")
con.execute("SET memory_limit='2GB'")
con.execute("SET threads=2")

print("Parquet files ready.")
```

```python
con.execute("""
CREATE OR REPLACE TABLE april AS
SELECT
    client_hash_id,
    content_hash_id,
    SUM(gsc_impressions) AS apr_impressions,
    SUM(gsc_clicks) AS apr_clicks,
    COUNT(DISTINCT report_date) AS apr_days
FROM read_parquet(?)
WHERE gsc_data_available = TRUE
GROUP BY client_hash_id, content_hash_id
""", [apr_path])

print("April aggregation complete.")
print(con.execute("SELECT COUNT(*) FROM april").fetchone()[0], "client-content pairs")
```

```python
con.execute("""
CREATE OR REPLACE TABLE may AS
SELECT
    client_hash_id,
    content_hash_id,
    SUM(gsc_impressions) AS may_impressions,
    SUM(gsc_clicks) AS may_clicks,
    COUNT(DISTINCT report_date) AS may_days
FROM read_parquet(?)
WHERE gsc_data_available = TRUE
GROUP BY client_hash_id, content_hash_id
""", [may_path])

print("May aggregation complete.")
print(con.execute("SELECT COUNT(*) FROM may").fetchone()[0],
      "client-content pairs")
```

```python
con.execute("""
CREATE OR REPLACE TABLE june AS
SELECT
    client_hash_id,
    content_hash_id,
    SUM(gsc_impressions) AS jun_impressions,
    SUM(gsc_clicks) AS jun_clicks,
    COUNT(DISTINCT report_date) AS jun_days
FROM read_parquet(?)
WHERE gsc_data_available = TRUE
GROUP BY client_hash_id, content_hash_id
""", [jun_path])

print("June aggregation complete.")
print(
    con.execute("SELECT COUNT(*) FROM june").fetchone()[0],
    "client-content pairs"
)
```

```python
con.execute("""
CREATE OR REPLACE TABLE model_data AS
WITH joined AS (
    SELECT
        a.client_hash_id,
        a.content_hash_id,
        a.apr_impressions,
        a.apr_clicks,
        a.apr_days,
        m.may_impressions,
        m.may_clicks,
        m.may_days,
        j.jun_impressions,
        j.jun_clicks,
        j.jun_days
    FROM april a
    INNER JOIN may m
        USING (client_hash_id, content_hash_id)
    INNER JOIN june j
        USING (client_hash_id, content_hash_id)
),
features AS (
    SELECT *,
        apr_impressions + may_impressions AS hist_impressions,
        apr_clicks + may_clicks AS hist_clicks,
        apr_days + may_days AS hist_days,
        (apr_impressions + may_impressions)
            / NULLIF(apr_days + may_days, 0) * 30
            AS historical_30d_impressions,
        jun_impressions / NULLIF(jun_days, 0) * 30
            AS june_30d_impressions
    FROM joined
)
SELECT *,
    CASE
        WHEN june_30d_impressions < 0.80 * historical_30d_impressions
        THEN 1 ELSE 0
    END AS is_declining
FROM features
WHERE apr_days > 0
  AND may_days > 0
  AND hist_impressions >= 100
  AND hist_days >= 40
  AND jun_days >= 20
""")

print("Modeling dataset created.")

print(con.execute("""
SELECT
    COUNT(*) AS eligible_items,
    SUM(is_declining) AS declining_items,
    COUNT(*) - SUM(is_declining) AS non_declining_items,
    ROUND(100.0 * AVG(is_declining), 2) AS decline_percentage
FROM model_data
""").df())
```

```python
con.execute("""
CREATE OR REPLACE TABLE training_data AS
SELECT
    client_hash_id,
    hist_impressions,
    hist_clicks,
    hist_days,
    historical_30d_impressions,
    is_declining
FROM model_data
""")

print(con.execute("""
SELECT
    COUNT(*) AS rows,
    COUNT(DISTINCT client_hash_id) AS clients,
    SUM(is_declining) AS declining,
    COUNT(*) - SUM(is_declining) AS non_declining
FROM training_data
""").df())
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import GroupShuffleSplit
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, accuracy_score


train_df = con.execute("""
    SELECT * FROM training_data
""").df()

features = [
    "hist_impressions",
    "hist_clicks",
    "hist_days",
    "historical_30d_impressions"
]

X = train_df[features]
y = train_df["is_declining"].astype(int)
groups = train_df["client_hash_id"]

splitter = GroupShuffleSplit(
    n_splits=5,
    test_size=0.25,
    random_state=123
)

results = []

for split_no, (train_idx, test_idx) in enumerate(
    splitter.split(X, y, groups), start=1
):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    models = {
        "Logistic Regression": make_pipeline(
            StandardScaler(),
            LogisticRegression(max_iter=1000, random_state=42)
        ),
        "Random Forest": RandomForestClassifier(
            n_estimators=200,
            min_samples_leaf=10,
            random_state=42,
            n_jobs=2
        )
    }

    for model_name, model in models.items():
        model.fit(X_train, y_train)
        probabilities = model.predict_proba(X_test)[:, 1]
        predictions = (probabilities >= 0.5).astype(int)


        top_n = min(100, len(y_test))
        top_indices = np.argsort(probabilities)[-top_n:]
        precision_at_100 = y_test.iloc[top_indices].mean()

        results.append({
            "split": split_no,
            "model": model_name,
            "test_rows": len(test_idx),
            "test_clients": groups.iloc[test_idx].nunique(),
            "roc_auc": roc_auc_score(y_test, probabilities),
            "accuracy": accuracy_score(y_test, predictions),
            "precision_at_100": precision_at_100,
            "test_decline_rate": y_test.mean()
        })

results_df = pd.DataFrame(results)

print("Results for each split:")
display(results_df.round(4))

print("\nAverage results across five splits:")
display(
    results_df.groupby("model")[
        ["roc_auc", "accuracy", "precision_at_100", "test_decline_rate"]
    ].mean().round(4)
)
```

```python
results_df.to_csv(
    "/content/model_comparison_five_splits.csv",
    index=False
)

average_results = (
    results_df.groupby("model")[
        ["roc_auc", "accuracy", "precision_at_100", "test_decline_rate"]
    ]
    .mean()
    .reset_index()
)

average_results.to_csv(
    "/content/model_comparison_summary.csv",
    index=False
)

print("Saved detailed and average model results.")
display(average_results.round(4))
```

This project investigates how machine learning can prioritize content pages whose search performance is declining.

**Research question:** Can a Random Forest model rank content pages labeled as declining more effectively than a Logistic Regression baseline?

**Decision supported:** Help human reviewers decide which pages to investigate first when review time is limited.

The model supports prioritization only. It does not prove why a page declined or guarantee that updating it will improve search performance.

### Dataset

This project uses the FlyRank internship warehouse full release hosted on Hugging Face.

### Tables used

- `dim_clients` — client-level metadata and data-history information.
- `dim_content` — content-level metadata.
- `fact_content_daily_performance` — daily search performance data.
- `fact_content_query_90d` — aggregated content/query-level signals.

### Data preparation and privacy

The analysis combines historical content performance with query-level signals to create a modeling dataset.

Client names, domains, URLs, search queries, and other identifying information are excluded from the public-facing analysis. Pseudonymous client identifiers are used only to separate clients between the training and test sets, not as model features.

### Date windows and leakage risk

The decline label compares impressions in the previous 30-day period with impressions in the latest 30-day period. Query-level signals come from the available 90-day query table. Because this window may overlap with the outcome period, potential temporal leakage is a limitation of the analysis.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

### Target definition

A content item is labeled as declining when its latest 30-day impressions are less than 80% of its previous 30-day impressions, representing a decline of more than 20%.

### Features

The model uses five features:

- `imp_prev30` — previous 30-day impressions.
- `visible_queries` — visible query count.
- `rare_share` — rare-query impression share.
- `anon_share` — anonymized-query impression share.
- `top_query_share` — top-query impression share.

## 5. Limitations

*What this work cannot claim.*

This experiment has several limitations:

1. **Potential temporal leakage:** The 90-day query-level signals may overlap with the outcome period.
2. **Limited generalization evidence:** The test set contains 13 clients. Performance on other clients or future periods may differ.
3. **Lower overall accuracy:** The Random Forest achieved approximately 61.3% accuracy, compared with 65.8% for Logistic Regression.
4. **No causal evidence:** The model identifies patterns associated with the declining-content label. It does not prove why content performance declined.
5. **No guaranteed improvement:** The model has not established that refreshing a page will improve its search performance.
6. **Experimental status:** The results are based on one holdout evaluation and should not be treated as proof of production readiness.

Further work should test temporally separated data, investigate leakage, and evaluate performance on additional clients and future periods.
